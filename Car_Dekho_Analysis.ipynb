```python
data["Depreciation"] = data["Present_Price"] - data["Selling_Price"]
data["Depreciation"].describe()
```

```python
data.loc[data["Depreciation"].idxmax()]
```

```python
data.loc[data["Depreciation"].idxmin()]
```

```python
data["Depreciation_Percentage"] = (
    (data["Present_Price"] - data["Selling_Price"])
    / data["Present_Price"]
) * 100

data[[
    "Car_Name",
    "Present_Price",
    "Selling_Price",
    "Depreciation",
    "Depreciation_Percentage"
]].sort_values(
    "Depreciation_Percentage",
    ascending=False
)
```

```python
data["Vehicle_Age"] = 2018 - data["Year"]
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=data,
    x="Vehicle_Age",
    y="Selling_Price"
)

plt.title("Vehicle Age vs Selling Price")
plt.xlabel("Vehicle Age")
plt.ylabel("Selling Price (₹ lakh)")

plt.tight_layout()
plt.savefig("charts/age_vs_price.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=data,
    x="Kms_Driven",
    y="Selling_Price"
)

plt.title("Kilometers Driven vs Selling Price")
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (₹ lakh)")

plt.tight_layout()
plt.savefig("charts/km_vs_price.png")
plt.show()
```

```python
year_count = data["Year"].value_counts().sort_index()

plt.figure(figsize=(10,5))

year_count.plot(kind="bar")

plt.title("Number of Vehicles by Manufacturing Year")
plt.xlabel("Year")
plt.ylabel("Number of Vehicles")

plt.tight_layout()
plt.savefig("charts/year_distribution.png")
plt.show()
```

```python
bike_keywords = [
    "Activa", "Bajaj", "Hero", "Honda", "Hyosung",
    "KTM", "Mahindra Mojo", "Royal Enfield",
    "Suzuki Access", "TVS", "UM Renegade", "Yamaha"
]

data["Vehicle_Type"] = np.where(
    data["Car_Name"].str.contains(
        "|".join(bike_keywords),
        case=False,
        na=False
    ),
    "Two-Wheeler",
    "Car"
)
```

```python
data["Vehicle_Type"].value_counts()
```

```python
bikes = data[data["Vehicle_Type"] == "Two-Wheeler"]
data.loc[bikes["Year"].idxmin(),['Car_Name','Year']]
```

```python
data.loc[bikes["Year"].idxmax(),['Car_Name','Year']]
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=data,
    x="Present_Price",
    y="Selling_Price",
    hue="Vehicle_Type"
)

plt.title("Present Price vs Selling Price")
plt.xlabel("Present Price (₹ lakh)")
plt.ylabel("Selling Price (₹ lakh)")

plt.tight_layout()
plt.savefig("charts/present_vs_selling.png")
plt.show()
```
